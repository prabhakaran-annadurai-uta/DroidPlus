```python
import pandas as pd
```

```python
df = pd.read_csv("inference_outputs/metrics.csv")
```

```python
df
```

```python
#mean lift time for successful lifts
mean_lift_time = df[df['outcome'] == 'success']['duration_sec'].mean()
print(f"Mean lift time for successful lifts: {mean_lift_time:.2f} seconds")
```

```python
#mean lift time for successful lifts grouped by run_label
mean_lift_time_by_label = df[df['outcome'] == 'success'].groupby('run_label')['duration_sec'].mean()
print(mean_lift_time_by_label)
```

```python
#success rate for each run_label
success_rate = df.groupby('run_label')['outcome'].apply(lambda x: (x == 'success').mean()).reset_index(name='success_rate')
print(success_rate)
```
